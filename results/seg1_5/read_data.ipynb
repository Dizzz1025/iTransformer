```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
pred = np.load('pred.npy')
true = np.load('true.npy')
```

```python
print('pred shape:', pred.shape)
print('true shape:', true.shape)
```

```python
pred_sample = pred.squeeze()
true_sample = true.squeeze()
```

```python
print('pred_sample shape:', pred_sample.shape)
print('true_sample shape:', true_sample.shape)
```

```python
# 画图
plt.figure(figsize=(10, 4))
plt.plot(true_sample, label='True')
plt.plot(pred_sample, label='Predicted')
plt.title(f'true vs pred')
plt.xlabel('Time Step')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```
